# Turkish Crowdfunding Campaign Success Classification

TURKISH CROWDFUNDING CAMPAIGN SUCCESS CLASSIFICATION


## 1. Understanding the Classification Problem

### Classification problem

The aim is to predict the **success status of a crowdfunding project** using information describing the project, platform, campaign design, project owner and promotional characteristics. The target variable is `basari_durumu`.

This is a **binary classification problem** because the dataset contains two target classes:

- `başarılı` = successful
- `başarısız` = unsuccessful

### Input variables

The dataset contains numerical and categorical predictors. Examples include:

- Platform: `platform_adi`
- Crowdfunding type: `kitle_fonlamasi_turu`
- Project category: `kategori`
- Funding model: `fon_sekli`
- Project-owner gender: `proje_sahibi_cinsiyet`
- Location and region: `konum`, `bolge`
- Campaign duration: `gun_sayisi`
- Promotional video availability and length: `tanitim_videosu`, `video_uzunlugu`
- Number of visuals: `gorsel_sayisi`
- Rewards: `odul_sayisi`
- Team size: `ekip_kisi_sayisi`
- Website/social-media indicators and follower counts
- Number of tags and description word count
- Funding target: `hedef_miktari`

Some variables are deliberately excluded from predictive modelling because they would cause **target leakage**. In particular, `toplanan_tutar` (amount collected) and `destek_orani` (support/funding percentage) directly reveal whether the funding target was achieved. Variables generated after or during the campaign such as supporter count, comments and updates are also excluded from the main launch-time prediction model.

### Output variable

`basari_durumu` is encoded as:

- 0 = unsuccessful
- 1 = successful

### Practical significance

A reliable model could help crowdfunding platforms and project creators assess campaign risk before launch. It could support decisions about project design, campaign duration, funding target, promotional material and marketing effort.

### Consequences of classification errors

A **false negative** occurs when a campaign that would succeed is predicted to fail. This could discourage a creator from launching, reduce marketing support, or lead the platform to overlook a promising campaign.

A **false positive** occurs when an unsuccessful project is predicted to succeed. This could lead to misplaced promotional resources, unrealistic expectations or poor funding decisions.

Because successful projects are less common in this dataset, accuracy alone may hide poor performance on the minority class. Therefore, this notebook reports **balanced accuracy, precision, recall, F1-score and ROC-AUC** in addition to standard accuracy.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split, StratifiedKFold, GridSearchCV, cross_validate
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay,
    RocCurveDisplay, classification_report
)

RANDOM_STATE = 42
```

### Data-quality observations and missing-value strategy

The dataset contains **1,628 rows and 38 columns**. The most substantial missingness occurs in the raw start and end date fields, while `bolge` contains only a very small amount of missing data.

For modelling, raw project start/end dates are removed because `gun_sayisi` already provides an interpretable campaign-duration variable. Removing the raw date strings also avoids unnecessary date parsing and duplicated information.

For remaining numerical predictors, missing values are imputed using the **median**. Median imputation is robust to skewed numerical distributions and extreme values. For categorical predictors, the **most frequent category** is used because it preserves the categorical representation without inventing numerical values.

A limitation of simple imputation is that it does not model why values are missing. If missingness itself contains predictive information, a future study could add explicit missing-value indicators.

```python
# Inspect target and key dataset evidence
print("Target values:")
print(df["basari_durumu"].value_counts())

print("\nMissing values (non-zero only):")
missing = df.isna().sum()
print(missing[missing > 0].sort_values(ascending=False))
```

Preventing target leakage
The purpose of the model is to make a meaningful prediction rather than reconstruct the final campaign outcome from variables that directly contain it. Therefore:

toplanan_tutar is excluded because it is the final amount collected.
destek_orani is excluded because it directly compares achieved funding with the target.
destekci_sayisi, yorumlar, and guncellemeler are excluded from the main model because they develop during the campaign and would not necessarily be known at launch.
IDs, project names, owner names and raw descriptions are excluded because they are identifiers or high-cardinality free text. The dataset already provides icerik_kelime_sayisi, which captures description length numerically.

```python
target_map = {"başarısız": 0, "başarılı": 1}
y = df["basari_durumu"].map(target_map)

drop_columns = [
    "basari_durumu",
    "id",
    "proje_adi",
    "proje_sahibi",
    "proje_aciklamasi",
    "proje_baslama_tarihi",
    "proje_bitis_tarihi",
    # Outcome / post-launch leakage variables
    "toplanan_tutar",
    "destek_orani",
    "destekci_sayisi",
    "yorumlar",
    "guncellemeler"
]

X = df.drop(columns=drop_columns)

continuous_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

print("Number of modelling features:", X.shape[1])
print("\nContinuous/numerical features:")
print(continuous_features)
print("\nCategorical features:")
print(categorical_features)
```

## Train-test split and preprocessing

A 70:30 stratified train-test split is used. Stratification preserves approximately the same success/failure proportion in both partitions, which is important because the target is imbalanced.

Preprocessing is fitted only on the training data through a scikit-learn `Pipeline`, preventing information from the test set leaking into training.

- Numerical variables: median imputation + standardisation.
- Categorical variables: most-frequent imputation + one-hot encoding.
- Unknown categories at test time are safely ignored by the encoder.

Standardisation is particularly useful for Logistic Regression because features measured on very different scales can otherwise affect optimisation and coefficient magnitude. Random Forest does not require scaling, but using one common preprocessing pipeline keeps the comparison reproducible.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    stratify=y,
    random_state=RANDOM_STATE
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)
print("\nTraining class proportions:")
print(y_train.value_counts(normalize=True).round(3))
print("\nTest class proportions:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, continuous_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

### 3. Model Development and Selection
Two different models are compared:

Model 1: Logistic Regression
Logistic Regression is selected as a strong interpretable baseline. It estimates the probability of success using a linear combination of transformed features.

Assumptions and strengths: it assumes an approximately linear relationship between predictors and the log-odds of the target, is computationally efficient, and produces interpretable coefficients.

Weaknesses: it may underfit complex nonlinear relationships and interactions unless those relationships are engineered explicitly.

Model 2: Random Forest
Random Forest is an ensemble of decision trees. It is selected because crowdfunding success may depend on nonlinear relationships and interactions among campaign characteristics.

Strengths: flexible nonlinear decision boundaries, reduced variance compared with a single tree, and robustness to complex interactions.

Weaknesses: higher computational cost and lower direct interpretability than Logistic Regression. Very deep forests can also fit noise.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

logistic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=3000, random_state=RANDOM_STATE))
])

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        random_state=RANDOM_STATE,
        n_jobs=-1
    ))
])
```

## Baseline models before imbalance handling

```python
def evaluate_model(model, X_eval, y_eval, model_name):
    pred = model.predict(X_eval)
    prob = model.predict_proba(X_eval)[:, 1]

    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_eval, pred),
        "Balanced Accuracy": balanced_accuracy_score(y_eval, pred),
        "Precision": precision_score(y_eval, pred, zero_division=0),
        "Recall": recall_score(y_eval, pred, zero_division=0),
        "F1": f1_score(y_eval, pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_eval, prob)
    }

baseline_models = {
    "Logistic Regression": logistic_pipeline,
    "Random Forest": rf_pipeline
}

baseline_results = []

for name, model in baseline_models.items():
    model.fit(X_train, y_train)
    baseline_results.append(evaluate_model(model, X_test, y_test, name))

baseline_results_df = pd.DataFrame(baseline_results).set_index("Model")
display(baseline_results_df.round(4))
```

#Hyperparameter optimisation
For Logistic Regression, the main tuned hyperparameter is C, the inverse regularisation strength. Smaller values impose stronger regularisation and can reduce overfitting; larger values allow a more flexible fit.

For Random Forest, the search explores the number of trees, tree depth, minimum samples needed to split a node, minimum leaf size, and number of features considered at each split. These parameters control the bias-variance and computational-cost trade-off.

The searches use five-fold stratified cross-validation and optimise ROC-AUC, because ROC-AUC evaluates probability ranking across classification thresholds and is less misleading than plain accuracy when classes are imbalanced

```python
logistic_param_grid = {
    "model__C": [0.01, 0.1, 1, 10, 100]
}

logistic_search = GridSearchCV(
    logistic_pipeline,
    logistic_param_grid,
    scoring="roc_auc",
    cv=cv,
    n_jobs=-1,
    return_train_score=True
)

logistic_search.fit(X_train, y_train)

print("Best Logistic Regression parameters:", logistic_search.best_params_)
print("Best CV ROC-AUC:", round(logistic_search.best_score_, 4))
```

```python
rf_param_grid = {
    "model__n_estimators": [150, 300],
    "model__max_depth": [None, 8, 15],
    "model__min_samples_split": [2, 5],
    "model__min_samples_leaf": [1, 3],
    "model__max_features": ["sqrt", 0.5]
}

rf_search = GridSearchCV(
    rf_pipeline,
    rf_param_grid,
    scoring="roc_auc",
    cv=cv,
    n_jobs=-1,
    return_train_score=True
)

rf_search.fit(X_train, y_train)

print("Best Random Forest parameters:", rf_search.best_params_)
print("Best CV ROC-AUC:", round(rf_search.best_score_, 4))
```

### Model diagnosis: underfitting, overfitting or generalising well

Training and validation ROC-AUC are compared using the best hyperparameters. A large training-validation gap suggests overfitting; low scores on both sets suggest underfitting; similar and reasonably strong scores suggest useful generalisation.

```python
def diagnostic_summary(search, name):
    best_idx = search.best_index_
    train_auc = search.cv_results_["mean_train_score"][best_idx]
    val_auc = search.cv_results_["mean_test_score"][best_idx]
    return {
        "Model": name,
        "Mean Train ROC-AUC": train_auc,
        "Mean Validation ROC-AUC": val_auc,
        "Train-Validation Gap": train_auc - val_auc
    }

diagnostics = pd.DataFrame([
    diagnostic_summary(logistic_search, "Logistic Regression"),
    diagnostic_summary(rf_search, "Random Forest")
]).set_index("Model")

display(diagnostics.round(4))
```

### 4. Addressing Data Imbalance
The dataset contains considerably more unsuccessful than successful projects, so a classifier could achieve deceptively high accuracy by favouring the majority class.

The selected imbalance-handling technique is class weighting (class_weight='balanced'). This gives greater training importance to the minority class without creating synthetic crowdfunding observations. It is suitable here because it is simple, reproducible and avoids assumptions required by synthetic oversampling methods.

The effect is evaluated by comparing the original tuned models against class-weighted versions.

```python
balanced_logistic = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        C=logistic_search.best_params_["model__C"],
        class_weight="balanced",
        max_iter=3000,
        random_state=RANDOM_STATE
    ))
])

balanced_rf = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=rf_search.best_params_["model__n_estimators"],
        max_depth=rf_search.best_params_["model__max_depth"],
        min_samples_split=rf_search.best_params_["model__min_samples_split"],
        min_samples_leaf=rf_search.best_params_["model__min_samples_leaf"],
        max_features=rf_search.best_params_["model__max_features"],
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=-1
    ))
])

balanced_logistic.fit(X_train, y_train)
balanced_rf.fit(X_train, y_train)

imbalance_results = []

imbalance_results.append(
    evaluate_model(logistic_search.best_estimator_, X_test, y_test,
                   "Logistic - tuned, original")
)
imbalance_results.append(
    evaluate_model(balanced_logistic, X_test, y_test,
                   "Logistic - class weighted")
)
imbalance_results.append(
    evaluate_model(rf_search.best_estimator_, X_test, y_test,
                   "Random Forest - tuned, original")
)
imbalance_results.append(
    evaluate_model(balanced_rf, X_test, y_test,
                   "Random Forest - class weighted")
)

imbalance_results_df = pd.DataFrame(imbalance_results).set_index("Model")
display(imbalance_results_df.round(4))
```

### 5. Final Model Evaluation and Recommendation

The final model is selected using cross-validation evidence together with test-set ROC-AUC, balanced accuracy, F1 and recall. The untouched test set is used only after model selection/tuning.

The following cells compare candidate models, inspect errors, plot ROC curves and then repeat the experiment using an 80:20 split to evaluate robustness

```python
candidate_models = {
    "Tuned Logistic Regression": logistic_search.best_estimator_,
    "Balanced Logistic Regression": balanced_logistic,
    "Tuned Random Forest": rf_search.best_estimator_,
    "Balanced Random Forest": balanced_rf
}

final_comparison = []
for name, model in candidate_models.items():
    final_comparison.append(evaluate_model(model, X_test, y_test, name))

final_comparison_df = pd.DataFrame(final_comparison).set_index("Model")
display(final_comparison_df.sort_values("ROC-AUC", ascending=False).round(4))
```

```python
# Choose the model with the highest ROC-AUC on this experimental comparison.
# If practical minority-class recall is the primary deployment priority,
# the class-weighted alternative should also be considered.
best_name = final_comparison_df["ROC-AUC"].idxmax()
best_model = candidate_models[best_name]

print("Selected model based on highest test ROC-AUC:", best_name)

best_pred = best_model.predict(X_test)
best_prob = best_model.predict_proba(X_test)[:, 1]

print("\nClassification report:")
print(classification_report(
    y_test, best_pred,
    target_names=["Unsuccessful", "Successful"],
    digits=4
))
```

```python
plt.figure(figsize=(6, 5))
ConfusionMatrixDisplay.from_predictions(
    y_test,
    best_pred,
    display_labels=["Unsuccessful", "Successful"],
    values_format="d"
)
plt.title(f"Confusion Matrix - {best_name}")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
for name, model in candidate_models.items():
    RocCurveDisplay.from_estimator(
        model, X_test, y_test, name=name, ax=plt.gca()
    )
plt.plot([0, 1], [0, 1], linestyle="--")
plt.title("ROC Curves - 70:30 Split")
plt.tight_layout()
plt.show()
```

## Error analysis
The confusion matrix separates the model's mistakes into:

False positives: unsuccessful projects predicted as successful.
False negatives: successful projects predicted as unsuccessful.
For a crowdfunding screening tool, false negatives may be particularly important because a genuinely promising campaign could be denied support. A model with slightly lower accuracy but noticeably higher recall for successful projects may therefore be preferable depending on the deployment objective.

The deployment threshold could also be adjusted instead of using the default probability threshold of 0.5.

```python
X_train80, X_test20, y_train80, y_test20 = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=RANDOM_STATE
)

logistic_search80 = GridSearchCV(
    logistic_pipeline,
    logistic_param_grid,
    scoring="roc_auc",
    cv=cv,
    n_jobs=-1
)
logistic_search80.fit(X_train80, y_train80)

rf_search80 = GridSearchCV(
    rf_pipeline,
    rf_param_grid,
    scoring="roc_auc",
    cv=cv,
    n_jobs=-1
)
rf_search80.fit(X_train80, y_train80)

results_80_20 = pd.DataFrame([
    evaluate_model(
        logistic_search80.best_estimator_,
        X_test20, y_test20,
        "Logistic Regression"
    ),
    evaluate_model(
        rf_search80.best_estimator_,
        X_test20, y_test20,
        "Random Forest"
    )
]).set_index("Model")

display(results_80_20.round(4))
```
